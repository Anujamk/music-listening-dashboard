```python
clean.groupby("spotify_track_uri")[["track_name", "artist_name", "album_name"]].nunique().max()
```

```python
counts = clean.groupby("spotify_track_uri")[["track_name", "album_name"]].nunique()
bad_uris = counts[(counts > 1).any(axis=1)].index
print(len(bad_uris))

clean[clean["spotify_track_uri"].isin(bad_uris)] \
    .drop_duplicates(["spotify_track_uri", "track_name", "album_name"]) \
    .sort_values("spotify_track_uri").head(20)
```
